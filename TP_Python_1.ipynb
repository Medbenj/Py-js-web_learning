2.1 — Programme convertisseur.py : 
convertir une température Celsius en Fahrenheit et inversement. L’utilisateur choisit le sens via un menu simple (mais on n’utilise pas encore if — on demande juste deux entrées).

```python
#convertisseur.py
Temp_C = input("entrer la temperature en Celsius : ")
Temp_F = input("entrer la temperature en Fahrenheit : ")

Temp_C_F = (float(Temp_C) * 9/5) + 32
Temp_F_C = (float(Temp_F) - 32) * 5/9

print("La temperature en Fahrenheit est : " + str(Temp_C_F))
print("La temperature en Celsius est : " + str(Temp_F_C))
```

5.3 — vitesse.py : lire une distance (km) et une durée (h, min, s). Calculer la vitesse moyenne en km/h avec deux décimales.

```python
#vitesse.py
Distance = int(input("entrer la distance en Km : "))
heures = int(input("entrer les heures  : "))
minutes = int(input("entrer les minutes  : "))
secondes = int(input("entrer les seconde  : "))
Duree = heures + minutes / 60 + secondes / 3600
vitesse = Distance / Duree
print(f"Le vitesse est : {vitesse:.2f} Km/h")
```
